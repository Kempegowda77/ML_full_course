```python
import pandas as pd
import re

df = pd.read_csv("railway_dataset_v2.csv")

df['Class_Type'] = df['Class'].str.extract(r'([A-Za-z]+)')
df['Class_Number'] = df['Class'].str.extract(r'(\d+)')

df['Seat_Number'] = df['Seat_Code'].apply(
    lambda x: re.search(r'\d+', str(x)).group() if re.search(r'\d+', str(x)) else None
)

df['Seat_Letter'] = df['Seat_Code'].apply(
    lambda x: re.search(r'[A-Za-z]+', str(x)).group() if re.search(r'[A-Za-z]+', str(x)) else None
)

print(df.head(10))
```
